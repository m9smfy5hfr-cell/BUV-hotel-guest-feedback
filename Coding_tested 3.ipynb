```python
### the store
feedbacks = []

### option 1:
def give_feedback():
    ## greeting
    print("=============================================")
    print("           ---- GIVE FEEDBACK ----")
    guest_name = input("\n- Guest name:")
    print("Please rate from 1 - 5 ")
    
    ## input
    while True: 
        Cleanliness = (input("\n- Cleanliness:")).strip()
        if Cleanliness.isdigit():
            Cleanliness = int(Cleanliness)
            if 1 <= Cleanliness <= 5:
                break
            else:
                print("Invalid input! Please rating from 1 - 5!")
        else: 
            print("Invalid input! Please rating from 1 - 5!")
            

    while True: 
        Room_quality = input("\n- Room quality:").strip()
        if Room_quality.isdigit():
            Room_quality = int(Room_quality)
            if 1 <= Room_quality <= 5:
                break
            else:
                print("Invalid input! Please rating from 1 - 5!")
        else: 
            print("Invalid input! Please rating from 1 - 5!")
            
    
    while True: 
        Staff_service = input("\n- Staff service:").strip()
        if Staff_service.isdigit():
            Staff_service = int(Staff_service)
            if 1 <= Staff_service <= 5:
                break
            else:
                print("Invalid input! Please rating from 1 - 5!")
        else: 
            print("Invalid input! Please rating from 1 - 5!")
   
        
    while True: 
        Price = input("\n- Price:").strip()
        if Price.isdigit():
            Price = int(Price)
            if 1 <= Price <= 5:
                break
            else:
                print("Invalid input! Please rating from 1 - 5!")
        else: 
            print("Invalid input! Please rating from 1 - 5!")
                
    
    while True: 
        Location = input("\n- Location:").strip()
        if Location.isdigit():
            Location = int(Location)
            if 1 <= Location <= 5:
                break
            else:
                print("Invalid input! Please rating from 1 - 5!")
        else: 
            print("Invalid input! Please rating from 1 - 5!")
            
    
    
    ## average value:
    average = (Location + Price + Staff_service + Room_quality + Cleanliness) / 5 
    print(f"\n- The overall satisfaction feedback: {average}")
    if average >= 4.5 :
        level = "Excellent Feedback"
    elif average >= 3.5:
        level = "Average Feedback"
    else:
        level = "Poor Feedback"

    print("- Feedback types:", level)
    print("\n                Thank You")
    print("=============================================")
    print("\n\n\n\n")

    
    ## store value
    feedback = {"Guest Name": guest_name,
                "Cleanliness": Cleanliness, 
                "Room quality": Room_quality,
                "Staff service": Staff_service,
                "Price":Price,
                "Location":Location,
                "average":average,
                "level": level}
    feedbacks.append(feedback)


### option 2:
def View_feedback():
    print("\n=============================================")
    print("        ---- ALL GUEST FEEDBACK ----")
    if len(feedbacks) == 0:
        print("Oops! Look like there are no feedback yet! Please try again!")
        return
    for feedback in feedbacks:
        print("=============================================")
        print("\n- Guest Name:", feedback["Guest Name"])
        criteria = ["Cleanliness","Room quality","Staff service","Price","Location"]
        for i in criteria:
            print(f"- {i}: {feedback[i]}/5")
        print("- The feedback overall:",round(feedback["average"],2))
        print("- The feedback level:",feedback["level"])
        print("\n                 The End")
        print("=============================================")
        print("\n\n\n\n")


### option 3:
def Feedback_report():
    print("\n=============================================")
    print("         ---- FEEDBACK REPORT ----")
    if len(feedbacks) == 0:
        print("Oops! Look like there are no feedback yet! Please try again!")
        return
    ## First call:
    Cleanliness_total = 0
    Room_quality_total = 0
    Staff_service_total = 0
    Price_total = 0
    Location_total = 0
    overall_total = 0
    
    excellent_level = 0
    average_level = 0
    poor_level = 0

    
    
    ## Total scores:
    for feedback in feedbacks:
        Cleanliness_total = Cleanliness_total + feedback["Cleanliness"]
        Room_quality_total = Room_quality_total + feedback["Room quality"]
        Staff_service_total = Staff_service_total + feedback["Staff service"]
        Price_total = Price_total + feedback["Price"]
        Location_total = Location_total + feedback["Location"]
        overall_total = overall_total + feedback["average"]

        ## Count feedbacks by classification
        if feedback["level"] == "Excellent Feedback":
            excellent_level = excellent_level + 1
        elif feedback["level"] == "Average Feedback":
            average_level = average_level + 1
        else:
            poor_level = poor_level + 1

    ## Number:
    number = len(feedbacks)

    ## Average scores
    Cleanliness_average = Cleanliness_total / number
    Room_quality_average = Room_quality_total / number
    Staff_service_average = Staff_service_total / number
    Price_average = Price_total / number
    Location_average = Location_total / number

    overall_average = overall_total/ number

    ## show out the average:
    print("\n")
    print("1.The average value:")
    print("- Overall average satisfaction:", round(overall_average, 2))
    average = {"Cleanliness": Cleanliness_average,
               "Room quality": Room_quality_average,
               "Staff service": Staff_service_average,
               "Price": Price_average,
               "Location": Location_average}
               
    for key in average:
        print(f"- {key} average: {round(average[key],1)}")
    ## show the classification:
    print("\n")
    print("2.Feedback classification:")
    print("- Excellent feedback number:", excellent_level)
    print("- Average feedback number:", average_level)
    print("- Poor feedback number:", poor_level)

    ##The Most or Least
    highest = lowest = round(average["Cleanliness"],1)
    strongest = weakest = "Cleanliness"

    for item in average:
        if average[item] > highest:
            highest = average[item]
            strongest = item
        if average[item] < lowest:
            lowest = average[item]
            weakest = item 
            
    print("\n")
    print("3.The specific area")
    print(f"- The Strongest Area is: {strongest} with {highest}/5")
    print(f"- The Area need to improve most is: {weakest} with {lowest}/5")
    
    print("\n                 The End")
    print("=============================================")
    print("\n\n\n\n")
        
    
### Menu:
while True:
    print("=============================================")
    print("               Sunrise Hotel")
    print("           Guest feedback system")
    print("=============================================")
    print("\n1. Give feedback")
    print("2. View feedback")
    print("3. Feedback report")
    print("4. Exit")

    choice = input("\nChoose:").strip()

    if choice == "1":
        give_feedback()
    elif choice == "2":
        View_feedback()
    elif choice == "3":
        Feedback_report()
    elif choice == "4":
        print("\n --- Thank you for using the service! ---")
        print("=============================================")
        break 
    else:
        print("Invalid value! Please select again!")
```
